Hai alias, mỗi FK nối đúng một dòng catalog.

```python
read_sql('SELECT f.flight_id,o.code origin,d.code destination FROM flights f JOIN airports o ON o.airport_id=f.origin_id JOIN airports d ON d.airport_id=f.dest_id LIMIT 20')
```

So số dòng toàn bộ trước/sau JOIN.

```python
read_sql('SELECT (SELECT COUNT(*) FROM flights) source_count,COUNT(*) joined_count FROM flights f JOIN airports o ON o.airport_id=f.origin_id JOIN airports d ON d.airport_id=f.dest_id')
```
