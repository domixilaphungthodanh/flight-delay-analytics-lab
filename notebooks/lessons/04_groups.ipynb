HAVING lọc nhóm sau tổng hợp; không chỉ nhìn số chuyến trễ.

```python
read_sql("SELECT c.reporting_code,COUNT(*) total,SUM(eligible) eligible,SUM(status='delayed') delayed FROM flight_status f JOIN carriers c USING(carrier_id) GROUP BY c.carrier_id,c.reporting_code HAVING SUM(eligible)>=1000")
```
